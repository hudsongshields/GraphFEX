```python
import pandas as pd

snr_levels = [30, 45, 60]
HR_fex_smapes = list(reversed([0.011587, 0.258068, 1.377027]))
Lorenz_fex_smapes = list(reversed([0.007000787597725743, 0.16296467509479837, 4.474004358155896]))
# Create a DataFrame to store the SNR levels and corresponding FEX sMAPE values (each row should just be a series of the corresponding smapes for that snr level)
fex_df = pd.DataFrame({
    'SNR': snr_levels,
    'smapes': [[HR_fex_smapes[i]/100, Lorenz_fex_smapes[i]/100] for i in range(len(snr_levels))]
})

HR_scindy_smapes = [16.056085742169948, 3.8690776529165825, 0.8521754253334978]
Lorenz_scindy_smapes = [0.28918069440828326, 0.005000250012497853, 0.0]

sindy_df = pd.DataFrame({
    'SNR': snr_levels,
    'smapes': [[HR_scindy_smapes[i]/100, Lorenz_scindy_smapes[i]/100] for i in range(len(snr_levels))]
})
```
